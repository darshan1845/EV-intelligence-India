```python
X=df.drop(['energy_consumed_kwh'],axis=1)
Y=df['energy_consumed_kwh']
```

```python
## One hot encoding for categorical  features
numeric_feature=X.select_dtypes(exclude='object').columns
OneHot_columns=['vehicle_type', 'charging_type', 'weather_condition', 'user_income_level']

from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer

numeric_transformer=StandardScaler()
Onehot_transformer=OneHotEncoder()

preprocessor=ColumnTransformer(
    [
        ("OneHotEncoder",Onehot_transformer,OneHot_columns),
        ("StandardScaler",numeric_transformer,numeric_feature)
    ]
)
```

```python
X=preprocessor.fit_transform(X)
```

```python
pd.DataFrame(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.20,random_state=42)
```

```python
X_train.shape
X_test.shape
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
```

```python
def evaluate_model(true,predicted):
    mse=mean_squared_error(true,predicted)
    mae=mean_absolute_error(true,predicted)
    rmse=np.sqrt(mse)
    R2_score=r2_score(true,predicted)
    return mae,rmse,R2_score
```

```python
models={
    "Random Forest Regression" : RandomForestRegressor(),
    "Linear Regression" : LinearRegression(),
    "Ridge" : Ridge(),
    "Lasso" : Lasso(),
    "XgBoost Regression" : XGBRegressor()
}

for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(X_train,Y_train)

   ##Prediction
    Y_train_pred=model.predict(X_train)
    Y_test_pred=model.predict(X_test)

   ##Calculate
    model_train_mae,model_train_rmse,model_train_r2=evaluate_model(Y_train,Y_train_pred)
    model_test_mae,model_test_rmse,model_test_r2=evaluate_model(Y_test,Y_test_pred)

    print(list(models.keys())[i])

    print("Model Performance for Training Data")
    print("- Root Mean Squared Error: {:.4f}".format(model_train_rmse))
    print("- Mean Absolute Error: {:.4f}".format(model_train_mae))
    print("- R2 Score: {:.4f}".format(model_train_r2))

    print('----------------------------------------')

    print("Model Performance for Test Set")
    print("- Root Mean Squared Error: {:.4f}".format(model_test_rmse))
    print("- Mean Absolute Error: {:.4f}".format(model_test_mae))
    print("- R2 Score: {:.4f}".format(model_test_r2))
    
    print('='*35)
    print('\n')
    
```

```python
##Initializing few parametes for hyperparameter tuning 
rf_params= {"max_depth": [5, 8, 15, None, 10],
             "max_features": [5, 7, "auto", 8],
             "min_samples_split": [2, 8, 15, 20],
             "n_estimators": [100, 200, 500, 1000]}

xgb_params= {"learning_rate": [0.1, 0.01],
                  "max_depth": [5, 8, 12, 20, 30],
                  "n_estimators": [100, 200, 300],
                  "colsample_bytree": [0.5, 0.8, 1, 0.3, 0.4]}
```

```python
rf_params
```

```python
## Models list for hyperparameter tuning
randomcv_models= [
    ("RF",RandomForestRegressor(),rf_params),
    ("XGB",XGBRegressor(),xgb_params)
    ]
```

```python
## Hyperparameter tuning
from sklearn.model_selection import RandomizedSearchCV

model_param={}
for name,model,params in randomcv_models:
    random = RandomizedSearchCV(estimator=model,
                                 param_distributions=params,
                                   n_iter=100,
                                   cv=3,
                                   verbose=2,
                                   n_jobs=-1)

    random.fit(X_train, Y_train)
    model_param[name]=random.best_params_

for model_name in model_param:
    print(f"---------------- Best Params for {model_name} -------------------")
    print(model_param[model_name])
```

```python
models={
     
    "Random Forest Regressor": RandomForestRegressor(n_estimators= 1000, min_samples_split=  8, max_features=8 , max_depth= 15),
    "Xgboost Regressor": XGBRegressor(learning_rate=0.01 , max_depth=5 , n_estimators=300 ,colsample_bytree=1 )
}
    
for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(X_train,Y_train)

    ##Prediction
    Y_train_pred=model.predict(X_train)
    Y_test_pred=model.predict(X_test)

    ##calculate
    model_train_mae,model_train_rmse,model_train_r2= evaluate_model(Y_train,Y_train_pred)
    model_test_mae,model_test_rmse,model_test_r2= evaluate_model(Y_test,Y_test_pred)

    print(list(models.keys())[i])

    print("Model Performance for Training Data")
    print("- Root Mean Squared Error: {:.4f}".format(model_train_rmse))
    print("- Mean Absolute Error: {:.4f}".format(model_train_mae))
    print("- R2 Score: {:.4f}".format(model_train_r2))

    print('----------------------------------------')

    print("Model Performance for Test Set")
    print("- Root Mean Squared Error: {:.4f}".format(model_test_rmse))
    print("- Mean Absolute Error: {:.4f}".format(model_test_mae))
    print("- R2 Score: {:.4f}".format(model_test_r2))

    print('='*35)
    print('\n')
    
```

```python


xgb_final = XGBRegressor(
    learning_rate=0.01,
    max_depth=5,
    n_estimators=300,
    colsample_bytree=1,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

xgb_final.fit(X_train, Y_train)
```

```python
Y_train_pred_xgb = xgb_final.predict(X_train)
Y_test_pred_xgb = xgb_final.predict(X_test)
```

```python
xgb_train_mae, xgb_train_rmse, xgb_train_r2 = evaluate_model(
    Y_train,
    Y_train_pred_xgb
)

xgb_test_mae, xgb_test_rmse, xgb_test_r2 = evaluate_model(
    Y_test,
    Y_test_pred_xgb
)
```

```python
print("XGBoost Regression - Final")
print("--------------------------------")

print("Training Data")
print("RMSE:", round(xgb_train_rmse, 4))
print("MAE :", round(xgb_train_mae, 4))
print("R²  :", round(xgb_train_r2, 4))

print()

print("Test Data")
print("RMSE:", round(xgb_test_rmse, 4))
print("MAE :", round(xgb_test_mae, 4))
print("R²  :", round(xgb_test_r2, 4))
```
